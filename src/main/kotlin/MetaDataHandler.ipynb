```kotlin
%useLatestDescriptors
%use dataframe
@file:DependsOn("com.github.doyaaaaaken:kotlin-csv-jvm:1.7.0")

import com.github.doyaaaaaken.kotlincsv.dsl.csvWriter
import io.github.oshai.kotlinlogging.KotlinLogging.logger
import kotlin.reflect.full.declaredMemberProperties

//avgClusterSize,avgClusters,avgClustersExcluded,avgClustersInPath,avgClustersIncludedCompletely,avgNodesInDeadCluster,avgPercentageBudgetUnused,avgPercentageClusterExcluded,avgPercentageClusterRevenueExcluded,avgRevenueInDeadCluster,budgetDist,budgetFactor,clustering,elimination,eliminationRevenueWeight,eliminationSparsityWeight,k,mode,r,runs,set

@DataSchema
data class OneRun(
    val avgClusterSize: Double,
    val avgClusters: Double,
    val avgClustersExcluded: Double,
    val avgClustersInPath: Double,
    val avgClustersIncludedCompletely: Double,
    val avgNodesInDeadCluster: Double,
    val avgPercentageBudgetUnused: Double,
    val avgPercentageClusterExcluded: Double,
    val avgPercentageClusterRevenueExcluded: Double,
    val avgRevenueInDeadCluster: Double,
    val budgetDist: String,
    val budgetFactor: Double,
    val clustering: String,
    val elimination: String,
    val eliminationRevenueWeight: Double,
    val eliminationSparsityWeight: Double,
    val k: Int,
    val mode: String,
    val r: Double,
    val runs: Int,
    val set: String
)

//avgClusterPercentageIncluded,avgClusterSize,avgClustersIncludedCompletely,avgNodesInDeadCluster,avgPercentageBudgetUnused,avgPercentageClusterExcluded,avgPercentageClusterRevenueExcluded,avgRevenueInDeadCluster,budgetDist,budgetFactor,clustering,elimination,eliminationRevenueWeight,eliminationSparsityWeight,k,mode,r,runs,set

data class globalRun(
    val avgClusterPercentageIncluded: Double,
    val avgClusterSize: Double,
    val avgClustersIncludedCompletely: Double,
    val avgNodesInDeadCluster: Double,
    val avgPercentageBudgetUnused: Double,
    val avgPercentageClusterExcluded: Double,
    val avgPercentageClusterRevenueExcluded: Double,
    val avgRevenueInDeadCluster: Double,
    val budgetDist: String,
    val budgetFactor: Double,
    val clustering: String,
    val elimination: String,
    val eliminationRevenueWeight: Double,
    val eliminationSparsityWeight: Double,
    val k: Int,
    val mode: String,
    val r: Double,
    val runs: Int,
    val set: String
)

val relativePath = "/op-solver-strict/results/meta/"
val pathToFolder  = java.nio.file.Paths.get(System.getProperty("user.dir"), "../../../..").normalize().toAbsolutePath()
    .toString() + relativePath


val folder = File(pathToFolder)

val csvFiles = folder.listFiles { file -> file.extension == "csv" } ?: arrayOf()
val dfs: Map<String, DataFrame<OneRun>> = csvFiles.filter { it.name != "global.csv" }.associate { file ->
    val name = file.nameWithoutExtension
    val dfRaw = DataFrame.readCsv(file)
        .convert("avgClusterSize") { it.toString().toDouble() }
        .convert("avgClusters") { it.toString().toDouble() }
        .convert("avgClustersExcluded") { it.toString().toDouble() }
        .convert("avgClustersInPath") { it.toString().toDouble() }
        .convert("avgClustersIncludedCompletely") { it.toString().toDouble() }
        .convert("avgNodesInDeadCluster") { it.toString().toDouble() }
        .convert("avgPercentageBudgetUnused") { it.toString().toDouble() }
        .convert("avgPercentageClusterExcluded") { it.toString().toDouble() }
        .convert("avgPercentageClusterRevenueExcluded") { it.toString().toDouble() }
        .convert("avgRevenueInDeadCluster") { it.toString().toDouble() }
        .convert("budgetDist") { it.toString() }
        .convert("budgetFactor") { it.toString().toDouble() }
        .convert("clustering") { it.toString() }
        .convert("elimination") { it.toString() }
        .convert("eliminationRevenueWeight") { it.toString().toDouble() }
        .convert("eliminationSparsityWeight") { it.toString().toDouble() }
        .convert("k") { it.toString().toInt() }
        .convert("mode") { it.toString() }
        .convert("r") { it.toString().toDouble() }
        .convert("runs") { it.toString().toInt() }
        .convert("set") { it.toString() }
        .cast<OneRun>()
    name to dfRaw
}

val globalFile = folder.listFiles { file -> file.name == "global.csv" }.firstOrNull()
    ?: throw IllegalStateException("Global CSV file not found")

var globalDf = DataFrame.readCsv(globalFile)
    .convert("avgClusterPercentageIncluded") { it.toString().toDouble() }
    .convert("avgClusterSize") { it.toString().toDouble() }
    .convert("avgClustersIncludedCompletely") { it.toString().toDouble() }
    .convert("avgNodesInDeadCluster") { it.toString().toDouble() }
    .convert("avgPercentageBudgetUnused") { it.toString().toDouble() }
    .convert("avgPercentageClusterExcluded") { it.toString().toDouble() }
    .convert("avgPercentageClusterRevenueExcluded") { it.toString().toDouble() }
    .convert("avgRevenueInDeadCluster") { it.toString().toDouble() }
    .convert("budgetDist") { it.toString() }
    .convert("budgetFactor") { it.toString().toDouble() }
    .convert("clustering") { it.toString() }
    .convert("elimination") { it.toString() }
    .convert("eliminationRevenueWeight") { it.toString().toDouble() }
    .convert("eliminationSparsityWeight") { it.toString().toDouble() }
    .convert("k") { it.toString().toInt() }
    .convert("mode") { it.toString() }
    .convert("r") { it.toString().toDouble() }
    .convert("runs") { it.toString().toInt() }
    .convert("set") { it.toString() }
    .cast<globalRun>()

globalDf
```
